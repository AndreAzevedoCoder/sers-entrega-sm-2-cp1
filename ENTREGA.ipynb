# Dataset 1: Appliances Energy Prediction (UCI)

Situação: uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

Amostra: 10% dos registros (energydata_SAMPLE.csv).

```python
df1.info()
```

```python
df1.describe()
```

```python
df1.columns
```

```python
# 2.	Renomeie Appliances para Consumo_Eletrodomesticos e simplifique o nome de pelo menos três atributos ambientais.
df1.rename(columns={'Appliances':'Consumo_Eletrodomesticos',
                    'T1': 'Temperatura_1',
                    'T2': 'Temperatura_2',
                    'T3': 'Temperatura_3'
                    }, inplace=True)
```

```python
df1.columns
```

```python
df1.head()
```

```python
# 3.	Determine o maior consumo de eletrodomésticos registrado na amostra.
maximo = df1['Consumo_Eletrodomesticos'].max()
print(f'Maior consumo de eletrodomésticos registrado na amostra: {maximo} Wh')
```

```python
# 4.	Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame
#     contendo somente os registros acima desse limite.
limiar_70 = 0.70 * maximo
print(f'Limiar (70% do máximo): {limiar_70} Wh')

df1_alto = df1[df1['Consumo_Eletrodomesticos'] > limiar_70]
df1_alto.head()
```

```python
# 5.	Conte os registros selecionados e calcule qual percentual da amostra eles representam.
qtd_alto = len(df1_alto)
percentual_alto = qtd_alto / len(df1) * 100
print(f'Registros com consumo acima do limiar: {qtd_alto}')
print(f'Percentual da amostra: {percentual_alto:.2f}%')
```

```python
# 6.	Calcule a temperatura média de T1 e crie outro DataFrame contendo simultaneamente
#     consumo acima de 70% do máximo e temperatura acima da média.
media_T1 = df1['Temperatura_1'].mean()
print(f'Temperatura média de T1: {media_T1:.2f} °C')

df1_alto_temp = df1[(df1['Consumo_Eletrodomesticos'] > limiar_70) &
                    (df1['Temperatura_1'] > media_T1)]
df1_alto_temp.head()
```

```python
# 7.	Compare os dois DataFrames.
qtd_alto_temp = len(df1_alto_temp)
percentual_alto_temp = qtd_alto_temp / len(df1) * 100
print(f'Somente consumo > 70% do máximo: {qtd_alto} registros ({percentual_alto:.2f}%)')
print(f'Consumo alto e T1 acima da média: {qtd_alto_temp} registros ({percentual_alto_temp:.2f}%)')
print(f'Diferença: {qtd_alto - qtd_alto_temp} registros')
```

Análise: o filtro de consumo acima de 70% do máximo (539 Wh) seleciona 22 registros, cerca de 1,1% da amostra, ou seja, picos de consumo são raros nessa residência. Exigindo também T1 acima da média (21,66 °C) sobram 8 registros (0,41%). Como as condições são combinadas com &, o segundo DataFrame é um subconjunto do primeiro. A maior parte dos picos ocorreu com temperatura igual ou abaixo da média, e o segundo critério serve para localizar só os casos de consumo alto em ambiente mais quente.

# Dataset 2: Steel Industry Energy Consumption (UCI)

Situação: a equipe de gestão de energia de uma indústria siderúrgica deseja localizar situações de consumo elevado e verificar se esses períodos coincidem com carga elevada ou com valores menos favoráveis de fator de potência.

Amostra: 20% dos registros (steel_SAMPLE.csv), com Usage_kWh, potências reativas, fatores de potência, WeekStatus, Day_of_week e Load_Type.

```python
# 1.	Carregue a amostra e renomeie Usage_kWh para Consumo_kWh,
#   	simplificando também os nomes dos atributos de fator de potência.
df2 = pd.read_csv('steel_SAMPLE.csv')
df2.rename(columns={'Usage_kWh': 'Consumo_kWh',
                    'Lagging_Current_Reactive.Power_kVarh': 'Pot_Reativa_Atraso_kVarh',
                    'Leading_Current_Reactive_Power_kVarh': 'Pot_Reativa_Avanco_kVarh',
                    'Lagging_Current_Power_Factor': 'FP_Atraso',
                    'Leading_Current_Power_Factor': 'FP_Avanco'}, inplace=True)
df2.columns
```

```python
# 2.	Inspeção inicial
df2.head()
```

```python
print(f'Total de registros: {df2.shape[0]}')
print(f'Total de colunas: {df2.shape[1]}')
```

```python
df2.info()
```

```python
df2.describe()
```

```python
# 3.	Determine o maior consumo registrado e calcule um limiar de 75% do consumo máximo.
maximo2 = df2['Consumo_kWh'].max()
limiar2_75 = 0.75 * maximo2
print(f'Maior consumo registrado: {maximo2} kWh')
print(f'Limiar (75% do máximo): {limiar2_75:.2f} kWh')
```

```python
# 4.	Crie um DataFrame com os registros acima do limiar e calcule quantidade e percentual.
df2_alto = df2[df2['Consumo_kWh'] > limiar2_75]
qtd2_alto = len(df2_alto)
pct2_alto = qtd2_alto / len(df2) * 100
print(f'Registros acima do limiar: {qtd2_alto}')
print(f'Percentual da amostra: {pct2_alto:.2f}%')
df2_alto.head()
```

```python
# 5.	Verifique quantos registros desse DataFrame pertencem à categoria Maximum Load.
print(df2_alto['Load_Type'].value_counts())
qtd2_max_load = len(df2_alto[df2_alto['Load_Type'] == 'Maximum_Load'])
print(f'\nRegistros de alto consumo na categoria Maximum_Load: {qtd2_max_load}')
```

```python
# 6.	Observe os valores de um atributo de fator de potência para definir um limite coerente.
df2['FP_Atraso'].describe()
```

```python
# 7.	DataFrame com consumo elevado E fator de potência abaixo do limite escolhido.
#   	Limite escolhido: FP_Atraso < 90 (abaixo de 90% o fator de potência é menos favorável).
limite_fp = 90
df2_critico = df2[(df2['Consumo_kWh'] > limiar2_75) & (df2['FP_Atraso'] < limite_fp)]
qtd2_critico = len(df2_critico)
print(f'Registros com consumo elevado e FP_Atraso < {limite_fp}: {qtd2_critico}')
print(f'Percentual da amostra: {qtd2_critico / len(df2) * 100:.2f}%')
df2_critico.head()
```

Análise: o limiar de 75% do consumo máximo (114,85 kWh) seleciona 129 registros (1,84%), sendo 62 da categoria Maximum_Load, quase metade. Pelo describe() do FP_Atraso, a mediana fica perto de 100, então 90 foi adotado como limite para fator de potência baixo. O cruzamento resultou em 42 registros com consumo alto e FP_Atraso abaixo de 90. Esses períodos merecem mais atenção porque o fator de potência baixo aumenta a circulação de potência reativa justamente quando a carga já está alta, elevando as perdas e podendo gerar cobrança de excedente de reativos. São os principais candidatos a correção, por exemplo com banco de capacitores.

# Dataset 3: Power Consumption of Tetouan City (UCI)

Situação: uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.

```python
# 1.	Carregar a amostra e renomear as três variáveis de consumo.
df3 = pd.read_csv('tetouan_SAMPLE.csv')
df3.rename(columns={'Zone 1 Power Consumption': 'Consumo_Zona_1',
                    'Zone 2  Power Consumption': 'Consumo_Zona_2',
                    'Zone 3  Power Consumption': 'Consumo_Zona_3'}, inplace=True)
df3.columns
```

```python
df3.head()
```

```python
print(f'Total de registros: {df3.shape[0]}')
print(f'Total de colunas: {df3.shape[1]}')
```

```python
# 6.	Calcule a temperatura média e crie um segundo DataFrame com consumo acima do
#   	limiar e temperatura acima da média.
media3_temp = df3['Temperature'].mean()
print(f'Temperatura média da amostra: {media3_temp:.2f} °C')
df3_alto_temp = df3[(df3[zona_pico] > limiar3_70) & (df3['Temperature'] > media3_temp)]
qtd3_alto_temp = len(df3_alto_temp)
print(f'Registros com consumo alto e temperatura acima da média: {qtd3_alto_temp}')
print(f'Percentual da amostra: {qtd3_alto_temp / len(df3) * 100:.2f}%')
```

# Dataset 5: Wind & Solar Energy Production (Kaggle)

Situação: um operador de um portfólio de fontes renováveis deseja comparar a ocorrência de períodos de alta produção solar e alta produção eólica. Como as duas fontes possuem escalas próprias, cada uma deve ser comparada com o seu próprio valor máximo.

Amostra: 20% dos registros (windsolar_SAMPLE.csv), com a produção horária solar e eólica em MW reorganizada em duas colunas a partir do atributo Source.

```python
# 1.	Renomeie as variáveis principais para Geracao_Solar e Geracao_Eolica.
df5 = pd.read_csv('windsolar_SAMPLE.csv')
df5.rename(columns={'Solar': 'Geracao_Solar', 'Wind': 'Geracao_Eolica'}, inplace=True)
df5.head()
```

```python
print(f'Total de registros: {df5.shape[0]}')
df5.describe()
```

```python
# 2.	Determine o valor máximo registrado para cada fonte.
max_solar = df5['Geracao_Solar'].max()
max_eolica = df5['Geracao_Eolica'].max()
print(f'Máximo solar:  {max_solar} MW')
print(f'Máximo eólico: {max_eolica} MW')
```

```python
# 3.	Calcule separadamente 70% do máximo solar e 70% do máximo eólico.
limiar_solar = 0.70 * max_solar
limiar_eolica = 0.70 * max_eolica
print(f'Limiar solar (70% do máximo solar):   {limiar_solar:.2f} MW')
print(f'Limiar eólico (70% do máximo eólico): {limiar_eolica:.2f} MW')
```

```python
# 4.	Crie um DataFrame para alta geração solar e outro para alta geração eólica.
df5_alta_solar = df5[df5['Geracao_Solar'] > limiar_solar]
df5_alta_eolica = df5[df5['Geracao_Eolica'] > limiar_eolica]
print(f'Registros de alta geração solar:  {len(df5_alta_solar)}')
print(f'Registros de alta geração eólica: {len(df5_alta_eolica)}')
```

```python
# 5.	Conte os registros de cada DataFrame e calcule seus respectivos percentuais.
pct_solar = len(df5_alta_solar) / len(df5) * 100
pct_eolica = len(df5_alta_eolica) / len(df5) * 100
print(f'Alta geração solar:  {len(df5_alta_solar)} registros ({pct_solar:.2f}% da amostra)')
print(f'Alta geração eólica: {len(df5_alta_eolica)} registros ({pct_eolica:.2f}% da amostra)')
```

```python
# 6.	Compare qual fonte aparece com maior frequência acima de 70% do seu próprio máximo.
fonte = 'eólica' if len(df5_alta_eolica) > len(df5_alta_solar) else 'solar'
print(f'A fonte {fonte} aparece com maior frequência acima de 70% do seu próprio máximo.')
```

Análise: comparando cada fonte com o próprio máximo, a eólica fica acima de 70% do seu máximo com mais frequência (296 registros, 4,95%) do que a solar (124 registros, 2,07%). Faz sentido, porque a solar só gera durante o dia e concentra os valores altos perto do meio-dia, enquanto a eólica pode operar perto da capacidade em qualquer horário. Não daria para usar o mesmo valor numérico como limite para as duas fontes: o máximo solar é 7.489 MW e o eólico 17.168 MW, então um limite de 12.018 MW nunca seria atingido pela solar. Por isso o limiar precisa ser relativo ao máximo de cada fonte.

# Dataset 6: Individual Household Electric Power Consumption (UCI)

Situação: uma residência possui monitoramento elétrico detalhado e deseja identificar episódios de demanda elevada que também apresentem corrente acima do comportamento médio.

Amostra: 10% do dataset completo (household_SAMPLE.csv), gerada após a remoção das linhas com valores ausentes e a seleção de Global_active_power, Global_reactive_power, Voltage, Global_intensity e os três Sub_metering.

```python
# 1.	Carregue a nova amostra e organize os atributos utilizando nomes mais simples.
df6 = pd.read_csv('household_SAMPLE.csv')
df6.rename(columns={'Global_active_power': 'Potencia_Ativa',
                    'Global_reactive_power': 'Potencia_Reativa',
                    'Voltage': 'Tensao',
                    'Global_intensity': 'Corrente',
                    'Sub_metering_1': 'Sub_Medicao_1',
                    'Sub_metering_2': 'Sub_Medicao_2',
                    'Sub_metering_3': 'Sub_Medicao_3'}, inplace=True)
df6.head()
```

```python
print(f'Total de registros: {df6.shape[0]}')
df6.info()
```

```python
# 2.	Determine o valor máximo de potência ativa.
maximo6 = df6['Potencia_Ativa'].max()
print(f'Potência ativa máxima: {maximo6} kW')
```

```python
# 3.	Calcule 75% do valor máximo e crie um DataFrame com registros acima desse limite.
limiar6_75 = 0.75 * maximo6
print(f'Limiar (75% do máximo): {limiar6_75:.3f} kW')
df6_alto = df6[df6['Potencia_Ativa'] > limiar6_75]
df6_alto.head()
```

```python
# 4.	Calcule a quantidade e o percentual de registros selecionados.
qtd6_alto = len(df6_alto)
pct6_alto = qtd6_alto / len(df6) * 100
print(f'Registros acima do limiar: {qtd6_alto}')
print(f'Percentual da amostra: {pct6_alto:.3f}%')
```

```python
# 5.	Calcule a corrente média da amostra.
media6_corrente = df6['Corrente'].mean()
print(f'Corrente média da amostra: {media6_corrente:.2f} A')
```

```python
# 6.	Segundo DataFrame: potência ativa acima de 75% do máximo E corrente acima da média.
df6_alto_corr = df6[(df6['Potencia_Ativa'] > limiar6_75) & (df6['Corrente'] > media6_corrente)]
qtd6_alto_corr = len(df6_alto_corr)
print(f'Registros com potência alta e corrente acima da média: {qtd6_alto_corr}')
print(f'Percentual da amostra: {qtd6_alto_corr / len(df6) * 100:.3f}%')
```

```python
# 7.	Compare os dois conjuntos.
print(f'Somente potência > 75% do máximo: {qtd6_alto} registros')
print(f'Potência alta e corrente acima da média: {qtd6_alto_corr} registros')
print(f'Diferença: {qtd6_alto - qtd6_alto_corr} registros')
```

Análise: acima de 75% do máximo de potência ativa (7,902 kW) há 58 registros (0,028% da amostra). Ao exigir também corrente acima da média (4,63 A), permanecem os mesmos 58, ou seja, a segunda condição não eliminou nenhum caso. Isso acontece porque a potência ativa é aproximadamente o produto da tensão pela corrente e a tensão varia pouco (perto de 240 V): manter mais de 7,9 kW exige uns 33 A, muito acima da média. Nesse limiar a condição de corrente é redundante e só filtraria algo com um limiar de potência mais baixo. Ao combinar condições com & vale checar se uma já implica a outra.
